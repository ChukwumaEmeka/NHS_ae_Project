```python
def clean_ae(filepath):
    """Load one monthly MSitAE CSV and return a cleaned provider-level DataFrame."""
    
    # 1. Load the raw file
    df = pd.read_csv(filepath)
    
    # 2. Clean column names, then apply rename_mapping
    df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('[^0-9a-zA-Z_]', '', regex=True)
    df = df.rename(columns=rename_mapping)
    
    # 3. Strip whitespace from the text columns
    text_cols = df.select_dtypes("object").columns
    for c in text_cols:
      
      df[c] = df[c].str.strip()

    
    # 4. Remove the England total row (labelled 'TOTAL' or 'Total' depending on the month)
    is_total = (df['org_code'].str.upper() == 'TOTAL') | (df['period'].str.upper() == 'TOTAL')
    df = df[~is_total]
    
    # 5. Convert period to datetime
    df["period"] = pd.to_datetime(df["period"].str.replace("MSitAE-", ""), format="%B-%Y")
    

    
    # 6. Add total_att and total_over_4hrs (use att_cols and over4_cols)
    att_cols = [
    'ae_att_type_1', 'ae_att_type_2', 'ae_att_others',
    'ae_att_type_1_booked', 'ae_att_type_2_booked', 'ae_att_others_booked'
]

    over4_cols = [
    'ae_att_over_4hrs_type_1', 'ae_att_over_4hrs_type_2', 'ae_att_over_4hrs_others',
    'ae_att_over_4hrs_type_1_booked', 'ae_att_over_4hrs_type_2_booked', 'ae_att_over_4hrs_others_booked'
    ]
    df['total_att'] = df[att_cols].sum(axis=1)
    df['total_over_4hrs'] = df[over4_cols].sum(axis=1)
    
    # 7. Add has_type1 flag
    df['has_type1'] = (df['ae_att_type_1'] + df['ae_att_type_1_booked']) > 0
    
    return df
```

```python
import glob

# List all monthly A&E files in date order
files = sorted(glob.glob('../data/raw/ae_*.csv'))
print(len(files))   # expect 24
files[:3]           # peek at the first three paths
```

```python
monthly = []                       # empty list to collect each cleaned month

for f in files:
    
    monthly.append(clean_ae(f))    # clean one file, add it to the list

ae_all = pd.concat(monthly, ignore_index=True)   # stack all 24 into one DataFrame
ae_all.shape
```

```python
# Some basic checks on the combined DataFrame
print(ae_all['period'].nunique())                 # expect 24
print((ae_all['org_code'].str.upper() == 'TOTAL').sum())   # expect 0
ae_all['period'].value_counts().sort_index()      # organisations per month
```

```python
#save the cleaned combined DataFrame to a CSV file for future use
ae_all.to_csv('../data/cleaned/ae_clean_2024_10_to_2026_09.csv', index=False)
```
